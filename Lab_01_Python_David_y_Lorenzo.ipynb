````python
#3
def consultar(cuenta):
    print(f"\n--- CONSULTA DE SALDO ---")
    print(f"Titular: {cuenta['titular']}")
    print(f"Saldo actual: ${cuenta['saldo']:,.0f}")


def depositar(cuenta, monto):
    if monto <= 0:
        print("El monto a depositar debe ser mayor a cero.")
        return False

    cuenta["saldo"] += monto
    # Registrar movimiento: (tipo, monto, saldo resultante)
    cuenta["movimientos"].append(("Depósito", monto, cuenta["saldo"]))
    print(f"Se han depositado ${monto:,.0f} correctamente.")
    return True


def retirar(cuenta, monto):
    if monto <= 0:
        print("El monto a retirar debe ser mayor a cero.")
        return False

    if monto > cuenta["saldo"]:
        print("No cuenta con el dinero suficiente, saldo negativo")
        return False

    cuenta["saldo"] -= monto
    cuenta["movimientos"].append(("Retiro", monto, cuenta["saldo"]))
    print(f"Se han retirado ${monto:,.0f} correctamente.")
    return True


# Estructura inicial de la cuenta
cuenta = {"titular": "Ana", "saldo": 200_000, "movimientos": []}

# Menú interactivo con while
while True:
    print("------ CAJERO AUTOMÁTICO ------")
    print("1. Consultar saldo")
    print("2. Depositar dinero")
    print("3. Retirar dinero")
    print("4. Salir")

    opcion = input("Seleccione una opción (1-4): ")

    if opcion == "1":
        consultar(cuenta)

    elif opcion == "2":
        try:
            monto_dep = float(input("Ingrese el monto a depositar: $"))
            depositar(cuenta, monto_dep)
        except ValueError:
            print("Ingrese un número válido.")

    elif opcion == "3":
        try:
            monto_ret = float(input("Ingrese el monto a retirar: $"))
            retirar(cuenta, monto_ret)
        except ValueError:
            print("Monto invalido, ingrese un número válido.")

    elif opcion == "4":
        print("Saliendo...")
        break
    else:
        print("Opción no válida. Intente de nuevo.")

# Al salir, muestra el extracto completo de movimientos en forma de tabla
print("                   EXTRACTO DE MOVIMIENTOS             ")
print()
print(f"Titular: {cuenta['titular']} | Saldo Final: ${cuenta['saldo']:,.0f}")
print("-------------------------------------------------------")
print(f"{'Tipo de Movimiento':<15} | {'Monto':<15} | {'Saldo Resultante':<15}")
print("-------------------------------------------------------")

if not cuenta["movimientos"]:
    print("No se registraron movimientos durante la sesión.")
else:
    for tipo, monto, saldo_res in cuenta["movimientos"]:
        print(f"{tipo:<15} | ${monto:>12,.0f}  | ${saldo_res:>14,.0f}")
````

````python
#2
import os
def guardar_csv(nombre_archivo, lista_productos):
    with open(nombre_archivo, "w", encoding="utf-8") as archivo:
        for p in lista_productos:
            linea = f"{p['nombre']},{p['precio']},{p['cantidad']}\n"
            archivo.write(linea)
def cargar_csv(nombre_archivo):
    lista_productos = []
    try:
        with open(nombre_archivo, "r", encoding="utf-8") as archivo:
            for linea in archivo:
                linea = linea.strip()
                if linea:
                    nombre, precio, cantidad = linea.split(",")
                    lista_productos.append({
                        "nombre": nombre,
                        "precio": float(precio),
                        "cantidad": int(cantidad)
                    })
    except FileNotFoundError:
        print("[Aviso] El archivo CSV no existe. Iniciando con un inventario vacío.")
    return lista_productos
```[cite: 4]


```python
import json

def guardar_json(nombre_archivo, lista_productos):
    with open(nombre_archivo, "w", encoding="utf-8") as archivo:
        json.dump(lista_productos, archivo, ensure_ascii=False, indent=4)
def cargar_json(nombre_archivo):
    try:
        with open(nombre_archivo, "r", encoding="utf-8") as archivo:
            return json.load(archivo)
    except FileNotFoundError:
        print("[Aviso] El archivo JSON no existe. Iniciando con un inventario vacío.")
        return []
```[cite: 4]

#### ¿Qué ventajas tiene JSON frente a CSV?
* **Estructura de datos nativa:** JSON preserva automáticamente los tipos de datos en Python (como enteros, flotantes, booleanos o diccionarios anidados). En CSV, todo se almacena como texto plano (`str`), lo que obliga a realizar conversiones manuales (`float()`, `int()`) al leerlo[cite: 4].
* **Menos complejidad con caracteres especiales:** Si los nombres de los productos contienen comas, comillas o saltos de línea, en un archivo CSV estándar romperían la estructura de columnas a menos que se apliquen reglas estrictas de escape. JSON maneja esto de manera transparente y robusta.
* **Facilidad de lectura y escritura:** Con una sola instrucción (`json.dump` o `json.load`), se puede serializar y deserializar estructuras complejas enteras sin necesidad de recorrer líneas y hacer divisiones manuales de cadenas (`split(",")`)[cite: 4].
````

````python
#punto #1
import random

def jugar():
    numero = random.randint(1, 100)
    historial = []
    intentos = 0
    ganado = False

    while intentos < 7 and ganado == False:
        intento = int(input("Adivina el numero: "))
        historial.append(intento)
        intentos = intentos + 1

        if intento == numero:
            print("¡Correcto! Ganaste.")
            ganado = True
        elif intento < numero:
            print("El numero es mayor.")
        else:
            print("El numero es menor.")

    if ganado == False:
        print("Perdiste. El numero era:", numero)

    print("Historial:", historial)

    return ganado


def main():
    ganadas = 0
    perdidas = 0
    jugar_otra = "s"

    while jugar_otra == "s":
        resultado = jugar()

        if resultado == True:
            ganadas = ganadas + 1
        else:
            perdidas = perdidas + 1

        print("Partidas ganadas:", ganadas)
        print("Partidas perdidas:", perdidas)

        jugar_otra = input("¿Deseas jugar otra vez? (s/n): ").lower()

main()
````

````python
# punto 2

def crear_tablero():
    tablero = []

    for i in range(10):
        fila = []
        for j in range(10):
            fila.append(0)
        tablero.append(fila)

    # Patron inicial
    tablero[1][2] = 1
    tablero[2][3] = 1
    tablero[3][1] = 1
    tablero[3][2] = 1
    tablero[3][3] = 1

    return tablero


def contar_vecinos(tablero, fila, columna):
    vecinos = 0

    for i in range(fila - 1, fila + 2):
        for j in range(columna - 1, columna + 2):

            if i >= 0 and i < 10 and j >= 0 and j < 10:
                if (i != fila or j != columna):
                    vecinos = vecinos + tablero[i][j]

    return vecinos


def siguiente_generacion(tablero):
    nuevo = []

    for i in range(10):
        fila = []

        for j in range(10):
            vecinos = contar_vecinos(tablero, i, j)

            if tablero[i][j] == 1:
                if vecinos == 2 or vecinos == 3:
                    fila.append(1)
                else:
                    fila.append(0)

            else:
                if vecinos == 3:
                    fila.append(1)
                else:
                    fila.append(0)

        nuevo.append(fila)

    return nuevo


def mostrar_tablero(tablero):
    for fila in tablero:
        for celda in fila:
            if celda == 1:
                print("■", end=" ")
            else:
                print(".", end=" ")
        print()


def main():
    tablero = crear_tablero()

    for i in range(5):
        print("Generacion:", i + 1)
        mostrar_tablero(tablero)
        tablero = siguiente_generacion(tablero)


main()
````

````python
# punto 1

def pedir_numero(mensaje, tipo):
    correcto = False

    while correcto == False:
        try:
            numero = tipo(input(mensaje))

            if numero < 0:
                print("El valor no puede ser negativo.")
            else:
                correcto = True

        except ValueError:
            print("Error. Ingresa un numero valido.")

        else:
            if correcto:
                return numero

        finally:
            pass


def agregar_producto(inventario):
    nombre = input("Nombre del producto: ")
    precio = pedir_numero("Precio: ", float)
    cantidad = pedir_numero("Cantidad: ", int)

    producto = {
        "nombre": nombre,
        "precio": precio,
        "cantidad": cantidad
    }

    inventario.append(producto)


def guardar(inventario):
    try:
        archivo = open("inventario.csv", "w")

        for producto in inventario:
            linea = producto["nombre"] + ";"
            linea = linea + str(producto["precio"]) + ";"
            linea = linea + str(producto["cantidad"]) + "\n"

            archivo.write(linea)

        archivo.close()

    except OSError:
        print("Error al guardar el archivo.")

    else:
        print("Inventario guardado correctamente.")

    finally:
        print("Proceso de guardado terminado.")


def cargar():
    inventario = []

    try:
        archivo = open("inventario.csv", "r")

        for linea in archivo:
            datos = linea.strip().split(";")

            producto = {
                "nombre": datos[0],
                "precio": float(datos[1]),
                "cantidad": int(datos[2])
            }

            inventario.append(producto)

        archivo.close()

    except FileNotFoundError:
        print("No existe el archivo. Inventario vacio.")

    return inventario


def mostrar(inventario):
    for producto in inventario:
        print(producto)


def main():
    inventario = cargar()
    opcion = 0

    while opcion != 4:
        print("\n1. Agregar producto")
        print("2. Mostrar inventario")
        print("3. Guardar inventario")
        print("4. Salir")

        opcion = int(input("Seleccione una opcion: "))

        if opcion == 1:
            agregar_producto(inventario)

        elif opcion == 2:
            mostrar(inventario)

        elif opcion == 3:
            guardar(inventario)

        elif opcion == 4:
            guardar(inventario)
            print("Programa terminado.")

        else:
            print("Opcion incorrecta.")


main()
````
